```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

#Heart disease analysis
This project shows the dataset of patients with indicators of heart diseases.
it includes data cleaning, visualization and insights necessary for a beginner health analytics portfolio

```python
url= "https://huggingface.co/datasets/buio/heart-disease/resolve/main/heart.csv"
df= pd.read_csv(url)
df.head()
```

##1. Age Distribution
Let's examine the age distribution of patients in this dataset.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=10, kde=True, color='skyblue')
plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

**Observation:**
Most patients between the ages of 55-60 are affected

##2. Gender Count
Let's look at the number of male and female patients in the dataset.
Note: females are represented with 0, males are represented with  1.

```python
plt.figure(figsize=(6,4))
sns.countplot(x="sex", data=df, palette="pastel")
plt.title("Gender Count (0=Female, 1=Male)")
plt.show()
```

**Observation:** Males patients are more than females in this dataset.

##3. Presence of Heart Disease
Now, let's look at how many patients have heart diseases in this dataset.
Note: 0= No heart disease, 1= Heart
 disease

```python
plt.figure(figsize=(6,4))
sns.countplot(x="target", data=df, palette="Set2")
plt.title("Presence of Heart Disease")
plt.show()
```

**Observation:** Less than a hundred patients have heart diisease in this dataset.


## 4. Age vs Max Heart Rate
We examine the relationship between age and maximum heart rate achieved, colored by heart disease presence.

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x="age",y="thalach", hue="target", data=df, palette="coolwarm")
plt.xlabel("Age")
plt.ylabel(" Max Heart Rate")
plt.title("Age vs Max Heart Rate")
plt.show()
```

**Observation:** Patients with heart disease generally have lower maximum heart rates, and older patients tend to show higher disease prevalence.

## Conclusion
This analysis provides a basic understanding of patient demographics and risk indicators for heart disease.  
The visualizations reveal trends in age, gender, heart rate, and disease prevalence, which can guide further studies or predictive modeling.
